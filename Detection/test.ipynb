```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
data = pd.read_csv('E:\Malicious-URL-detection\MaliciousURLDataset\MaliciousURLsTransformed.csv')
```

```python
x = data.drop(columns=['url_type'])
y = data['url_type'] 
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3, random_state=42)
```

```python
data.head()
```

```python
model = RandomForestClassifier()
model.fit(x_train, y_train)
```

```python
URL_encoding = {0: 'benign',
    1: 'defacement',
    2: 'phishing',
    3: 'malware'}

result = model.predict("https://drive.google.com/drive/u/0/my-drive")

print("\n\n\n\n\n RESULT\nType:"+URL_encoding[result])
```
